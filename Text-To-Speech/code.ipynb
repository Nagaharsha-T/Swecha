```python
from gtts import gTTS
from PIL import Image
import pytesseract
import pygame
import tempfile
import os

# Set path to Tesseract-OCR (Update this path if needed)
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

def text_to_speech_model():
    print("Starting TTS process...")

    # Step 1: Select Input Option
    input_option = input("Select input option \n 1. Manual Text Input \n 2. Image of Printed Text\n")

    if input_option == "1":
        text = input("Enter the text: ")
    elif input_option == "2":
        image_path = r"C:\Users\NAGAHARSHA\OneDrive\Documents\Text-To-Speech\Happiness.jpg"  # Update path
        text = pytesseract.image_to_string(Image.open(image_path))
        print(f"Extracted text from image:\n{text}")
    else:
        print("Invalid option. Please enter 1 or 2.")
        return

    # Step 2: Convert and Play Audio
    text_to_speech(text)

    # Step 3: Done
    print("TTS process completed.")

def text_to_speech(text, lang='en'):
    try:
        # Create temporary MP3 file
        with tempfile.NamedTemporaryFile(delete=False, suffix=".mp3") as temp_audio:
            audio_file = temp_audio.name

        # Use gTTS to convert text to speech
        tts = gTTS(text=text, lang=lang)
        tts.save(audio_file)
        print("Processing the text!! Please wait...")
        print("The text is as follows:")
        print(text)

        # Play the audio with pygame
        pygame.mixer.init()
        pygame.mixer.music.load(audio_file)
        pygame.mixer.music.play()

        # Wait until audio finishes playing
        while pygame.mixer.music.get_busy():
            pygame.time.Clock().tick(10)

        # Properly stop and delete file
        pygame.mixer.quit()
        os.remove(audio_file)

    except Exception as e:
        print(f"Error: {e}")

# Run the model
if __name__ == "__main__":
    text_to_speech_model()
```
